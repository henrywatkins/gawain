# Plotting example

```python
from gawain.io import HDFReader
```

```python
import numpy as np
import matplotlib.pyplot as plt

from matplotlib import animation, rc
from IPython.display import HTML
```

```python
data = Reader("../runs/mhd_rotor")
```

```python
data.variables
```

```python
data.plot("density", timesteps=[0, 10, 20, 23, 40, 98], vmin=0, vmax=12)
```

```python
raw_data = data.get_data("density")
```

```python
shape = (256, 256)
```

```python
fig, ax = plt.subplots(figsize=(20, 20))

image = ax.imshow(
    np.random.random(shape), vmin=raw_data.min(), vmax=raw_data.max(), cmap="plasma"
)
```

```python
# initialization function: plot the background of each frame
def init():
    image.set_data(np.random.random(shape))
    return (image,)
```

```python
# animation function. This is called sequentially
def animate(i):
    y = raw_data[i].reshape(shape)
    image.set_data(y)
    return (image,)
```

```python
anim = animation.FuncAnimation(
    fig, animate, init_func=init, frames=99, interval=10, blit=True
)
```

```python
HTML(anim.to_jshtml())
```

```python
data = HDFReader("../runs/sod_shock_tube.h5")
```

```python
data.plot("density", timesteps=[0, 20, 40, 60, 80, 98], vmin=0, vmax=1)
```

```python
raw_data = data.get_data("density")
```

```python
fig, ax = plt.subplots()

ax.set_xlim((0, 1))
ax.set_ylim((0, 10))

(line,) = ax.plot([], [], lw=2)
```

```python
# initialization function: plot the background of each frame
def init():
    line.set_data([], [])
    return (line,)
```

```python
# animation function. This is called sequentially
def animate(i):
    x = np.linspace(0, 1, 512)
    y = np.squeeze(raw_data[i])
    line.set_data(x, y)
    return (line,)
```

```python
anim = animation.FuncAnimation(
    fig, animate, init_func=init, frames=100, interval=20, blit=True
)
```

```python
HTML(anim.to_jshtml())
```
